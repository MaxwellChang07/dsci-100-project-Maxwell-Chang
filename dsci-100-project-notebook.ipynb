#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r

# Keep teams with valid seeds and the data needed for prediction.
# Seed 0 was a special 1983 designation, so exclude it.
tournament_data <- march_madness %>%
  filter(
    !is.na(seed), seed > 0,
    !is.na(reg_wins_pct),
    !is.na(tourney_wins)
  ) %>%

  # Turn tournament wins into a Yes/No categorical outcome.
  mutate(
    won_game = factor(
      if_else(tourney_wins > 0, "Yes", "No"),
      levels = c("No", "Yes")
    )
  ) %>%

  # Keep only predictors and the outcome used in the research question.
  select(seed, reg_wins_pct, won_game)

# Display the first six rows of the tidied data.
head(tournament_data)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
# Select every numeric column in the original data.
numeric_summary <- march_madness %>%
  select(
    year, seed,
    conf_wins:conf_rank,
    reg_wins:reg_wins_pct,
    tourney_wins:tourney_losses,
    total_wins:total_wins_pct
  ) %>%
  # Put the numeric measurements into one column for grouped summaries.
  pivot_longer(
    cols = year:total_wins_pct,
    names_to = "variable",
    values_to = "value"
  ) %>%
  group_by(variable) %>%
  summarize(
    mean = mean(value, na.rm = TRUE),
    minimum = min(value, na.rm = TRUE),
    maximum = max(value, na.rm = TRUE)
  )

# Display all 14 rows.
print(numeric_summary, n = 14)
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
# Count missing values before any rows or columns are removed.
missing_summary <- march_madness %>%
  summarize(
    across(year:total_wins_pct, ~ sum(is.na(.x)))
  ) %>%
  # Display one row per original variable.
  pivot_longer(
    cols = year:total_wins_pct,
    names_to = "variable",
    values_to = "missing_values"
  )

print(missing_summary, n = 20)
```

```r
options(repr.plot.width = 8, repr.plot.height = 5)

# Show the distribution of the numeric predictor.
win_pct_distribution <- tournament_data %>%
  ggplot(aes(x = reg_wins_pct)) +
  geom_histogram(
    binwidth = 5,
    fill = "steelblue",
    color = "white"
  ) +
  labs(
    title = "Distribution of regular-season win percentages",
    x = "Regular-season win percentage (%)",
    y = "Number of team appearances"
  ) +
  theme(text = element_text(size = 14))

win_pct_distribution
```

```r
# Count each outcome and convert its count to a percentage.
outcome_proportions <- tournament_data %>%
  group_by(won_game) %>%
  summarize(count = n()) %>%
  ungroup() %>%
  mutate(percentage = 100 * count / sum(count))

# Show the relative frequency of each outcome.
outcome_distribution <- outcome_proportions %>%
  ggplot(aes(x = won_game, y = percentage)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  labs(
    title = "Proportion of teams winning a tournament game",
    x = "Won at least one tournament game?",
    y = "Percentage of team appearances (%)"
  ) +
  theme(text = element_text(size = 14))

outcome_distribution
```

Visualization 1: This plot shows the percentage of teams that won at least one tournament game for each seed. Lower-numbered seeds generally had higher winning percentages, suggesting that seed may help predict tournament success.

```r
# Calculate the percentage winning at least once within each seed.
seed_win_rates <- tournament_data %>%
  group_by(seed) %>%
  summarize(
    win_percentage = 100 * sum(won_game == "Yes") / n()
  )

# Treat seeds as discrete labels for this plot.
seed_relationship <- seed_win_rates %>%
  ggplot(aes(x = as_factor(seed), y = win_percentage)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  labs(
    title = "Tournament winning percentage by seed",
    x = "Tournament seed",
    y = "Team appearances with at least one win (%)"
  ) +
  theme(text = element_text(size = 14))

seed_relationship
```

Visualization 2: These histograms compare regular-season win percentages for teams with and without a tournament win. Teams with a tournament win generally had higher regular-season win percentages, but the distributions still overlap. This suggests the variable may help predict tournament success but cannot perfectly distinguish the two outcomes.

```r
options(repr.plot.width = 8, repr.plot.height = 6)

# Compare the numeric predictor's distribution across outcome groups.
win_pct_relationship <- tournament_data %>%
  ggplot(aes(x = reg_wins_pct)) +
  geom_histogram(
    binwidth = 5,
    fill = "steelblue",
    color = "white"
  ) +
  facet_grid(rows = vars(won_game)) +
  labs(
    title = "Regular-season win percentage by tournament outcome",
    subtitle = "Yes = at least one tournament win; No = no tournament wins",
    x = "Regular-season win percentage (%)",
    y = "Number of team appearances"
  ) +
  theme(text = element_text(size = 14))

win_pct_relationship
```
